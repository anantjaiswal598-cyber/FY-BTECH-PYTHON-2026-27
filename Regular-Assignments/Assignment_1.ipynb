```python
mylist = ["red", "pink", "orange"]
print("initial colour to be selected :", mylist)
print("most likable colour :", mylist[1])
mylist[0] = "green"
print("new colours to be intoduces :", mylist)
mylist.append("blue")
print("all the colours :", mylist)
mylist.remove("pink")
print("the remaining part :", mylist)
mylist.sort()
print("in a sorted way :", mylist)
mylist.sort(reverse = True)
print("in the reverse order :", mylist)
mylist.copy()
print(mylist)
anotherlist = ["pen", "phone", "key"]
print("second list of objects :", anotherlist)
mylist.extend(anotherlist)
print("final set of lists :", mylist)
```
